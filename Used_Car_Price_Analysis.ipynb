```python

       import pandas as pd
import numpy as np

# Load your dataset (Replace 'cars.csv' with your file path or URL)
# df = pd.read_csv('your_dataset.csv')

# Load your dataset
df = pd.read_csv('/used_cars_messy.csv')

# Exploration
print("--- Dataset Info ---")
df.info()
print("\n--- Descriptive Statistics ---")
display(df.describe())
print(f"\nShape: {df.shape}")

# Identify 3 Issues (Observations to report):
# 1. Null values in 'selling_price' or feature columns.
# 2. 'mileage' is likely an object/string (e.g., '15 kmpl') instead of float.
# 3. 'brand' names might have inconsistent casing or leading spaces.
```

Task 1: Data Quality Issues Identified
After exploring the used_cars_messy.csv dataset, I identified the following three issues:
1.	Missing Values: The mileage and selling_price columns have significant null values (only 14,376 non-null rows out of 15,661).
2.	Incorrect Data Types: The mileage and km_driven columns are stored as objects (text) because they contain units like "kmpl" and commas. They must be numeric for analysis.
3.	Logical Errors: The seats column contains a minimum value of 0.0, which is an impossible value for a car and indicates a data entry error.

```python
# 1. Drop rows where selling_price is missing
df.dropna(subset=['selling_price'], inplace=True)

# 2. Extract numbers from 'mileage' and 'km_driven'
df['mileage'] = df['mileage'].astype(str).str.extract('(\d+\.\d+|\d+)').astype(float)
df['km_driven'] = df['km_driven'].astype(str).str.replace(',', '').str.extract('(\d+)').astype(float)

# 3. Impute (Fill) missing values
df['mileage'] = df['mileage'].fillna(df['mileage'].median())
df['engine'] = df['engine'].fillna(df['engine'].median())

# 4. Clean 'brand' column
df['brand'] = df['brand'].str.strip().str.lower()

# 5. Remove duplicates
df.drop_duplicates(inplace=True)

print("Cleaning complete! New Shape:", df.shape)
```

Task 2: Data Cleaning Summary
I performed the following cleaning steps:
* Dropped rows where the target variable (selling_price) was missing.
* Extracted numeric values from text columns using Regex and converted them to floats.
* Handled missing values in mileage and engine by imputing with the median.
* Standardized the brand column and removed duplicate rows.
* Final Dataset Shape: (15322, 14)

```python
from sklearn.metrics import mean_absolute_error

# Predict the average (mean) price for every car
y_true = df['selling_price']
prediction = y_true.mean()
y_pred = [prediction] * len(y_true)

# Calculate MAE
baseline_mae = mean_absolute_error(y_true, y_pred)

print(f"Mean Selling Price: {prediction:.2f}")
print(f"Baseline MAE: {baseline_mae:.2f}")
```
